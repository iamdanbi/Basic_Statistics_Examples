```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(42)
N = 100 # number of observations
# # grid X, not random
# x1_linear = np.linspace(0, 10, N)
# x2_linear = np.linspace(0, 10, N)
# X from uniform distribution
x1_linear = rng.uniform(0, 10, N)
x2_linear = rng.uniform(0, 10, N)
y_linear = 2*x1_linear + 7 * x2_linear + 3 + rng.normal(0, 0.1, size=N)

X_linear = np.column_stack([np.ones(N), x1_linear, x2_linear])

n_train = int(N * 0.8)
X_linear_train, X_linear_test = X_linear[:n_train, :], X_linear[n_train:, :]
y_linear_train, y_linear_test = y_linear[:n_train], y_linear[n_train:]
```

```python
beta_hat = np.linalg.inv(X_linear_train.T @ X_linear_train) @ X_linear_train.T @ y_linear_train
```
